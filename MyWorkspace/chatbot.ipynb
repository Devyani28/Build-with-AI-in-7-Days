```python
def getresponse(messages):
    response= client.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
        temperature= 0.7,
        
    )
    return response.choices[0].message.content
```

```python
messages = [
    {"role": "system", "content": " you are friendly and helpful assistant. Greet the user and answer the questions in a friendly manner."},
    {"role": "user", "content": "tell me about npci in 2   bullet points"}
]

response= getresponse(messages)
```

```python
messages.append({"role": "assistant", "content": response})
messages.append({"role": "user", "content": "what is the full form of that company"})
getresponse(messages)
```

```python
import gradio as gr

def getdata(message, chathistory):
    print(message)
    print(chathistory)

    messages = []

    for current_message in chathistory:
      messages.append(current_message)

    messages.append({"role": "user", "content": message})

    return getresponse(messages)

gr.ChatInterface(getdata).launch(debug=True)
```
