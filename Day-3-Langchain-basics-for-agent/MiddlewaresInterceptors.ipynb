```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model="gpt-4.1-mini")
```

```python
#tool
# Using the predefined LangChain Tavily tool
from  langchain_tavily import TavilySearch
# Create predefined Tavily tool
tavily_tool = TavilySearch(
    max_results=3,
    search_depth="advanced",
    include_answer=True,
    include_raw_content=False,
    include_images=False
)

print("Predefined Tavily tool created!")
```

```python
#logging interceptor b/w toolCalls & llmCalls
from langchain.agents.middleware import ModelRequest, wrap_model_call

#request have state, model/override(model=), 
#middleware for logging - 2interceptor calls(2logger we prints)
@wrap_model_call
def logging_middleware(request:ModelRequest,handler):
    messages= request.state["messages"]

    last_message=messages[-1]
    
    print(f"Logging Middleware: Last message role: {last_message}")
    response = handler(request)
    print(f"Logging Middleware: Last message role: {response}")

    return response
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    tools=[tavily_tool], #langchain api tools
    middleware=[logging_middleware]
)
```

```python
response = agent.invoke(
    {
        "messages":[
            {"role":"user","content":"tell me about devyani"}
        ]
    }
)
response
```
